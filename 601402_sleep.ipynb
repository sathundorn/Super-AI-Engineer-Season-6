```python
import pandas as pd
import os
import glob

# 1. กำหนด Path โฟลเดอร์ train ของคุณ
# (แก้โฟลเดอร์ให้ตรงกับที่คุณแตกไฟล์ sleepstage ไว้)
TRAIN_DIR = '/content/drive/MyDrive/sleepstage_data/super-ai-engineer-ss-6-sleep-stage-classification/train/train'

# 2. ดึงรายชื่อไฟล์ CSV ทั้งหมดในโฟลเดอร์
train_files = glob.glob(os.path.join(TRAIN_DIR, '*.csv'))

if len(train_files) > 0:
    print(f"✅ พบไฟล์ CSV ในโฟลเดอร์ train จำนวน {len(train_files)} ไฟล์")

    # 3. ลองอ่านไฟล์แรกสุดขึ้นมาดู
    sample_file = train_files[0]
    print(f"📄 กำลังเปิดอ่านไฟล์: {os.path.basename(sample_file)}")

    df_sample = pd.read_csv(sample_file)

    print("\n📊 สรุปข้อมูลในไฟล์:")
    print(f"-> จำนวนแถว (Time steps): {df_sample.shape[0]:,}")
    print(f"-> จำนวนคอลัมน์ (Features): {df_sample.shape[1]}")
    print("\nรายชื่อคอลัมน์ทั้งหมด:")
    print(df_sample.columns.tolist())

    print("\nตัวอย่างข้อมูล 5 แถวแรก:")
    display(df_sample.head())

else:
    print(f"❌ ไม่พบไฟล์ CSV ใน {TRAIN_DIR} โปรดตรวจสอบ Path อีกครั้งครับ")
```

```python
import pandas as pd
import numpy as np
import os
import glob
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.model_selection import train_test_split
import gc

feature_cols = ['BVP', 'ACC_X', 'ACC_Y', 'ACC_Z', 'TEMP', 'EDA', 'HR', 'IBI']
SEQ_LEN = 480 # 16 Hz * 30 วินาที

X_list = []
Y_list = []

train_files = glob.glob(os.path.join(TRAIN_DIR, '*.csv'))
print(f"⏳ กำลังประมวลผลไฟล์ทั้งหมด {len(train_files)} ไฟล์ (อาจใช้เวลา 2-3 นาที)...")

for idx, file_path in enumerate(train_files):
    df = pd.read_csv(file_path)

    # ดึงค่า Features และ Label ออกมาเป็น Numpy array เพื่อความรวดเร็ว
    x_data = df[feature_cols].values
    y_data = df['Sleep_Stage'].values

    # หั่นข้อมูลทีละ 480 แถว
    num_segments = len(df) // SEQ_LEN
    for i in range(num_segments):
        start = i * SEQ_LEN
        end = start + SEQ_LEN

        # เก็บสัญญาณ 480 แถว
        X_list.append(x_data[start:end])
        # เก็บ Label (โจทย์บอกว่าทั้ง 480 แถวใน 1 segment จะมี label เดียวกัน จึงดึงค่าแรกมาได้เลย)
        Y_list.append(y_data[start])

    if (idx + 1) % 10 == 0:
        print(f"   -> ประมวลผลไปแล้ว {idx + 1} ไฟล์...")

# แปลงเป็น Numpy Array ก้อนใหญ่
X = np.array(X_list)
Y = np.array(Y_list)

# ล้างหน่วยความจำ
del X_list, Y_list
gc.collect()

print(f"\n✅ หั่นข้อมูลเสร็จสิ้น! ได้ทั้งหมด {X.shape[0]:,} Segments")
print(f"📊 X shape: {X.shape} (Segments, Time steps, Features)")

# แปลง Label ตัวอักษร (W, N1, N2...) เป็นตัวเลข (0, 1, 2...)
le = LabelEncoder()
Y_encoded = le.fit_transform(Y)
print("📌 คลาสการนอนหลับที่พบ:", le.classes_)

# แปลงสเกลข้อมูล (Standardization) ให้ค่าแกว่งอยู่ในช่วงใกล้ๆ 0
print("⏳ กำลังปรับสเกลข้อมูล (Scaling)...")
scaler = StandardScaler()
# StandardScaler รับข้อมูลแค่ 2 มิติ จึงต้องคลาย (Flatten) ก่อนสเกล แล้วค่อยม้วนกลับ
X_scaled = scaler.fit_transform(X.reshape(-1, len(feature_cols))).reshape(X.shape)

# แบ่งข้อมูลสำหรับ Train (80%) และ Validation (20%)
X_train, X_val, y_train, y_val = train_test_split(X_scaled, Y_encoded, test_size=0.2, random_state=42)
print(f"-> แบ่งข้อมูล Train: {len(X_train)} ชุด | Validation: {len(X_val)} ชุด พร้อมเทรนแล้ว!")
```

```python
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import TensorDataset, DataLoader

# 1. ตั้งค่า GPU
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"กำลังรันบน: {device}")

# 2. แปลงข้อมูลเป็น Tensor (PyTorch ต้องการมิติ (Batch, Channels, Seq_Len) จึงต้องใช้ .permute)
X_train_t = torch.tensor(X_train, dtype=torch.float32).permute(0, 2, 1)
X_val_t = torch.tensor(X_val, dtype=torch.float32).permute(0, 2, 1)
y_train_t = torch.tensor(y_train, dtype=torch.long)
y_val_t = torch.tensor(y_val, dtype=torch.long)

# สร้าง DataLoader
train_loader = DataLoader(TensorDataset(X_train_t, y_train_t), batch_size=64, shuffle=True)
val_loader = DataLoader(TensorDataset(X_val_t, y_val_t), batch_size=64, shuffle=False)

# 3. โครงสร้างโมเดล 1D-CNN
class SleepCNN(nn.Module):
    def __init__(self, num_classes):
        super(SleepCNN, self).__init__()
        # Layer 1
        self.conv1 = nn.Conv1d(in_channels=8, out_channels=32, kernel_size=5, padding=2)
        self.relu1 = nn.ReLU()
        self.pool1 = nn.MaxPool1d(kernel_size=2) # ขนาดจาก 480 จะเหลือ 240

        # Layer 2
        self.conv2 = nn.Conv1d(in_channels=32, out_channels=64, kernel_size=5, padding=2)
        self.relu2 = nn.ReLU()
        self.pool2 = nn.MaxPool1d(kernel_size=2) # ขนาดจาก 240 จะเหลือ 120

        # Fully Connected Layer (Classifier)
        self.flatten = nn.Flatten()
        self.fc1 = nn.Linear(64 * 120, 128)
        self.relu3 = nn.ReLU()
        self.dropout = nn.Dropout(0.5) # ป้องกัน Overfitting
        self.fc2 = nn.Linear(128, num_classes)

    def forward(self, x):
        x = self.pool1(self.relu1(self.conv1(x)))
        x = self.pool2(self.relu2(self.conv2(x)))
        x = self.flatten(x)
        x = self.dropout(self.relu3(self.fc1(x)))
        x = self.fc2(x)
        return x

# 4. เรียกใช้โมเดล กำหนด Loss และ Optimizer
NUM_CLASSES = len(le.classes_)
model = SleepCNN(NUM_CLASSES).to(device)

criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters(), lr=0.005)

# 5. ลูปการเทรน
epochs = 15
print("\n🚀 เริ่มต้นเทรนโมเดล 1D-CNN...")

for epoch in range(epochs):
    # --- ท่อน Train ---
    model.train()
    running_loss = 0.0
    correct_train = 0

    for inputs, labels in train_loader:
        inputs, labels = inputs.to(device), labels.to(device)

        optimizer.zero_grad()
        outputs = model(inputs)
        loss = criterion(outputs, labels)
        loss.backward()
        optimizer.step()

        running_loss += loss.item()
        _, predicted = torch.max(outputs, 1)
        correct_train += (predicted == labels).sum().item()

    train_acc = 100 * correct_train / len(y_train)

    # --- ท่อน Validation ---
    model.eval()
    correct_val = 0
    with torch.no_grad():
        for inputs, labels in val_loader:
            inputs, labels = inputs.to(device), labels.to(device)
            outputs = model(inputs)
            _, predicted = torch.max(outputs, 1)
            correct_val += (predicted == labels).sum().item()

    val_acc = 100 * correct_val / len(y_val)

    print(f"Epoch [{epoch+1}/{epochs}] | Loss: {running_loss/len(train_loader):.4f} | Train Acc: {train_acc:.2f}% | Val Acc: {val_acc:.2f}%")

print("✅ เทรนโมเดลเสร็จสมบูรณ์!")
```

```python
import pandas as pd
import numpy as np
import os
import torch

# 1. กำหนด Path
# (ตรวจสอบให้แน่ใจว่าโฟลเดอร์ test_segment และไฟล์ submission อยู่ตรงนี้)
TEST_DIR = '/content/drive/MyDrive/sleepstage_data/super-ai-engineer-ss-6-sleep-stage-classification/test_segment/test_segment'
SUB_CSV_PATH = '/content/drive/MyDrive/sleepstage_data/super-ai-engineer-ss-6-sleep-stage-classification/sample_submission.csv'

df_sub = pd.read_csv(SUB_CSV_PATH)
id_col = df_sub.columns[0]
label_col = df_sub.columns[1]

# 2. ค้นหาไฟล์ .csv ทะลุทุกโฟลเดอร์ย่อย (subject000, subject001...)
print("🔍 กำลังสแกนหาไฟล์ .csv ทะลุทุกโฟลเดอร์ย่อย...")
all_test_files = glob.glob(os.path.join(TEST_DIR, '**', '*.csv'), recursive=True)

# สร้างแผนที่ (Dictionary) จับคู่ "ชื่อไฟล์" -> "Path เต็ม" เพื่อความรวดเร็ว
file_map = {}
for f in all_test_files:
    basename = os.path.basename(f)
    file_map[basename] = f
    file_map[basename.replace('.csv', '')] = f # เผื่อในไฟล์ submission ไม่มีนามสกุลติดมา

print(f"✅ สแกนพบไฟล์ .csv ทั้งหมด {len(all_test_files)} ไฟล์")
print(f"⏳ กำลังโหลดข้อมูลเข้าโมเดลตามคิวใน Submission จำนวน {len(df_sub)} ไฟล์...")

X_test_list = []
valid_indices = []

# 3. วนลูปอ่านไฟล์ตามรายชื่อใน Submission
for index, row in df_sub.iterrows():
    file_id = str(row[id_col])

    # ดึง Path เต็มจากแผนที่เราสร้างไว้
    file_path = file_map.get(file_id)

    if file_path and os.path.exists(file_path):
        df_test = pd.read_csv(file_path)
        x_data = df_test[feature_cols].values

        # ป้องกัน Error
        if len(x_data) > SEQ_LEN:
            x_data = x_data[:SEQ_LEN]
        elif len(x_data) < SEQ_LEN:
            x_data = np.pad(x_data, ((0, SEQ_LEN - len(x_data)), (0, 0)), mode='edge')

        X_test_list.append(x_data)
        valid_indices.append(index)
    else:
        print(f"⚠️ หาไฟล์ไม่พบ: {file_id}")

# 4. แปลงข้อมูล ทดสอบ และทำนายผล
if len(X_test_list) > 0:
    X_test_np = np.array(X_test_list)
    print("กำลังปรับสเกลข้อมูล Test...")
    X_test_scaled = scaler.transform(X_test_np.reshape(-1, len(feature_cols))).reshape(X_test_np.shape)

    X_test_t = torch.tensor(X_test_scaled, dtype=torch.float32).permute(0, 2, 1).to(device)

    print("🤖 โมเดลกำลังทำนายผล...")
    model.eval()
    predictions = []

    batch_size = 64
    with torch.no_grad():
        for i in range(0, len(X_test_t), batch_size):
            batch_inputs = X_test_t[i:i+batch_size]
            outputs = model(batch_inputs)
            _, preds = torch.max(outputs, 1)
            predictions.extend(preds.cpu().numpy())

    predicted_labels = le.inverse_transform(predictions)

    # นำคำตอบใส่ DataFrame ต้นแบบ
    for idx, pred_label in zip(valid_indices, predicted_labels):
        df_sub.loc[idx, label_col] = pred_label

    # 5. บันทึกไฟล์ส่ง
    output_filename = 'submission_sleepstage_cnn.csv'
    df_sub.to_csv(output_filename, index=False)
    print(f"\n✅ เสร็จสมบูรณ์! บันทึกไฟล์ '{output_filename}' เรียบร้อยแล้ว")
else:
    print("❌ ไม่พบข้อมูลให้ทำนายเลย ลองเช็ค Path หรือโฟลเดอร์อีกรอบครับ")
```
